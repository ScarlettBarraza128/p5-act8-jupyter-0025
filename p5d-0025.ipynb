```python
#  Dias y Temperatura
#  value lista de dias y lista de temperaturas 
datos = {
    "Dia": ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes"],
    "Temperatura": [27, 29, 31, 28, 30]
}
# DataFrame tabla o marco de datos
df = pd.DataFrame(datos)

print("=== TEMPERATURAS ===")
print(df)

fila = df["Temperatura"].idxmax()

print("Día con mayor temperatura:", df.loc[fila, "Dia"])
print("Temperatura máxima:", df["Temperatura"].max(), "°C")

# encuentra la posición donde está el valor máximo.
print("posición donde esta el valor máximo")
df.loc[fila, "Dia"]
print("-+-+-+-+-+-+-+-+-jupyter Emily+-+-+-+-+-+-+-")
print("-+-+-+-+-+-+-+-+-Emily Barraza NC 0025+-+-+-+-+-+-+-")
```
