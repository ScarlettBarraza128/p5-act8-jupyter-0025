```python
# productos de ventas de papeleria
# value lista de productos, lista de precio y lista de cantidad de producto

datos = {
    "Producto": ["Cuaderno", "Pluma", "Mochila", "Lápiz", "Regla"],
    "Precio": [35, 10, 250, 8, 15],
    "Cantidad": [5, 10, 2, 15, 6]
}

# DataFrame tabla o marco de datos
df = pd.DataFrame(datos)

df["Total"] = df["Precio"] * df["Cantidad"]

print("=== VENTAS DE PAPELERÍA ===")
print(df)

# Precio × Cantidad
print("Venta total:", df["Total"].sum())
print("-+-+-+-+-+-+-+-+-jupyter Emily+-+-+-+-+-+-+-")
print("-+-+-+-+-+-+-+-+-Emily Barraza NC 0025+-+-+-+-+-+-+-")
```
